11. ### Noticias NBC reportó el 2 de mayo de 2013 que 1 de cada 20 niños en los Estados Unidos presenta una alergia alimentaria de algún tipo. Considere que se selecciona una muestra aleatoria de 25 niños y sea X la cantidad de niños en la muestra que tienen una alergia alimentaria.

#### a) Determine P(X <3) y P(X ≤3)
#### b) Determine P(X ≥ 4)
#### c) Determine P(1 ≤ X ≤3).

```python
# Distribución binomial: n ensayos
n=25
p = 1/20
print(f'X~Bin(n={n}, p={p})')
print(f"X:=Número de niños que presentaban una alergia alimentaria en una muestra de 25 niños\n")

# a)
# P(X < 3) = P(X <= 2) = F(2)
a1 = dist.binom.cdf(2, n, p)
# P(X <= 3) = F(3)
a2 = dist.binom.cdf(3, n, p)

# b)
# P(X >= 4) = 1 - P(X <= 3) = 1 - F(3)
b = dist.binom.sf(3, n, p)

# c)
# P(1≤X≤3)=F(3)−F(0)
c = dist.binom.cdf(3, n, p) - dist.binom.cdf(0, n, p)

print(f"a1) P(X < 3) = {a1:.6f}")
print(f"a2) P(X <= 3) = {a2:.6f}")
print(f"b) P(X >= 4) = {b:.6f}")
print(f"c) P(1 <= X <= 3) = {c:.6f}")
```

### 12. Un componente eléctrico tiene una vida media de 8 años. Si su vida útil se distribuye en forma exponencial. ¿Cuál debe ser el tiempo de garantía que se debe otorgar, si se desea reemplazar a lo más el 15% de los componentes que fallen dentro de este periodo?

```python
# Distribución exponencial
media = 8
lambd = 1 / media
p = 0.15
print(f'X~Exp(lambda={lambd})')
print(f"X:=Tiempo de vida útil de un componente eléctrico, medido en años\n")

# P(X <= t) = 0.15 = F(t)
# F(t) = 1 - e^(-lambda*t)

# t = -ln(1-p) / lambda
t = -np.log(1 - p) / lambd

# En scipy, scale = 1/lambda = media
p12 = dist.expon.cdf(t, scale=media)

print(f"Tiempo de garantía: {t:.2f} años")
print(f"Equivalente en meses: {t*12:.2f}")
print(f"P(X <= {t:.2f}) = {p12:.6f}")

```

### 13.  El número de solicitudes de ayuda recibidas por un servicio de grúas es un proceso de Poisson con razón 4 llamadas por hora.
#### a) Calcule la probabilidad de que exactamente diez solicitudes sean recibidas durante un periodo particular de 2 horas.
#### b) Si los operadores del servicio de grúas hacen una pausa de 30 minutos para el almuerzo, ¿cuál es la probabilidad de que no dejen de atender las llamadas de auxilio?
#### c) ¿Cuántas llamadas esperaría durante esta pausa?

```python
# Distribución de Poisson: X ~ Poisson(lambda)
tasa = 4  # Llamadas por hora
print(f'X~Poisson(lambda)')
print(f"X:=Número de llamadas de solicitud de ayuda recibidas por el servicio de grúas durante un intervalo de tiempo determinado.\n")

print(f"Tasa: {tasa} llamadas por hora\n")

# a)
# P(X = 10) durante 2 horas
# lambda = 4 * 2 = 8
lambda_a = tasa * 2
a = dist.poisson.pmf(10, lambda_a)

# b)
# P(X = 0) durante 30 minutos
# No perder llamadas implica que no llegue ninguna
# lambda = 4 * 0.5 = 2
lambda_b = tasa * 0.5
b = dist.poisson.pmf(0, lambda_b)

# c)
# Número esperado de llamadas en 30 minutos
# E(X) = lambda
c = lambda_b

print(f"a) P(X = 10) = {a:.6f}")
print(f"b) P(X = 0) = {b:.6f}")
print(f"c) E(X) = {c:.0f} llamadas")
```

### 14. En una tienda departamental el tiempo promedio de espera para ser atendido en cajas al pagar la mercancía es de 7 minutos. Determine la probabilidad de que: 
#### a) Un cliente espere menos de 4 minutos. 
#### b) Un cliente espere más de 9 minuto

```python
# Distribución exponencial
# Tiempo promedio de espera en minutos
media = 7
lambd = 1 / media

print(f"X~Exp(lambda={lambd:.6f})")
print(f"X:=Tiempo de espera de un cliente (minutos)\n")

# a)
# P(X < 4) = P(X <= 4) = F(4)
#  P(X = 4) = 0
a = dist.expon.cdf(4, scale=media)

# b)
# P(X > 9) = 1 - P(X <= 9) = 1 - F(9)
b = 1 - dist.expon.cdf(9, scale=media)

print(f"a) P(X < 4) = {a:.6f}")
print(f"b) P(X > 9) = {b:.6f}")
```

### 15. Una lavadora MABE tiene una vida media de 10 años. Si la vida útil de ese motor puede consi- derarse como una variable aleatoria distribuida en forma exponencial. ¿Cuál debe ser el tiempo de garantía que deben tener dichas lavadoras si desea que a los más 20 % de estas fallen antes de que expire su garantía?

```python
# Distribución exponencial
# Vida media de la lavadora en años
media = 10
lambd = 1 / media
p = 0.20

print(f"X~Exp(lambda={lambd:.6f})")
print("X:=Tiempo de vida útil de una lavadora (años)\n")

# P(X <= t) = 0.20 = F(t)
# F(t) = 1 - e^(-lambda*t)
# Despejando: t = -ln(1-p) / lambda

t = -np.log(1 - p) / lambd

probabilidad = dist.expon.cdf(t, scale=media)

print(f"Tiempo de garantía = {t:.4f} años")
print(f"Equivalente = {t * 12:.2f} meses")
print(f"P(X <= {t:.4f}) = {probabilidad:.6f}")
```

### 16. Se supone que la estancia de los enfermos en un hospital sigue una distribución normal con media 8 días y desviación típica 3. Calcular la probabilidad de que la estancia de un enfermo: 
#### a) sea inferior a 7 días 
#### b) sea superior a 3 días 
#### c) esté comprendida entre 10 y 12 días.

```python
# Distribución normal
# Tiempo de estancia hospitalaria en días
mu = 8
sigma = 3

print(f"X~N(mu={mu}, sigma²={sigma**2})")
print("X:=Tiempo de estancia de un enfermo (días)\n")

# a)
# P(X < 7) = P(X <= 7) = F(7)
a = dist.norm.cdf(7, loc=mu, scale=sigma)

# b)
# P(X > 3) = 1 - P(X <= 3) = 1 - F(3)
b = 1 - dist.norm.cdf(3, loc=mu, scale=sigma)

# c)
# P(10 <= X <= 12) = F(12) - F(10)
c = (
    dist.norm.cdf(12, loc=mu, scale=sigma)
    - dist.norm.cdf(10, loc=mu, scale=sigma)
)

print(f"a) P(X < 7) = {a:.6f}")
print(f"b) P(X > 3) = {b:.6f}")
print(f"c) P(10 <= X <= 12) = {c:.6f}")
```

### 17.  Los salarios de los trabajadores de un país pueden suponerse que siguen una distribución normal de media 2000 euros y desviación típica desconocida. 
#### a) Si la probabilidad de ganar más de 2100 euros es de 0.33, ¿cuál es la desviación típica? 
#### b) Los salarios en euros de los trabajadores en un segundo país también puede suponerse que siguen una distribución normal con la misma media y con varianza de 40000 euros. ¿Es más fácil ganar más de 2100 euros en este segundo país que en el país del apartado anterior?

```python
# Distribución normal
mu = 2000
salario = 2100
print(f"X~N(mu={mu}, sigma²=?)")
print("X:=Salario de un trabajador (euros)\n")

# a)
# P(X1 > 2100) = 0.33
# P(X1 <= 2100) = 0.67 = F(2100)
# Z = (2100 - mu) / sigma

p = 0.33
# Obteniendo el valor z
z = brentq(lambda z: dist.norm.cdf(z) - (1-p), 0, 5)
sigma1 = (salario - mu) / z

# Comprobación: P(X1 > 2100) = 1 - F(2100)
a = 1 - dist.norm.cdf(salario, loc=mu, scale=sigma1)

# b)
# País 2: X2 ~ N(mu=2000, sigma²=40000)
# P(X2 > 2100) = 1 - F(2100)
varianza2 = 40000
sigma2 = varianza2 ** 0.5

b = 1 - dist.norm.cdf(salario, loc=mu, scale=sigma2)

print(f"X1~N(mu={mu}, sigma²={sigma1**2:.2f})")
print(f"X2~N(mu={mu}, sigma²={varianza2})\n")

print(f"a) Desviación típica = {sigma1:.4f} euros")
print(f"   P(X1 > 2100) = {a:.6f}")

print(f"\nb) P(X2 > 2100) = {b:.6f}")
print(f"   ¿Es más fácil en el país 2? {'Sí' if b > a else 'No'}")
```

### 18. La distribución de peso de paquetes enviados de cierta manera es normal con valor medio de 12 lb y desviación estándar de 3.5 lb. El servicio de paquetería desea establecer un valor de peso más allá del cual se hará un cargo extra. ¿Qué valor de es tal que el 99 % de todos los paquetes están al menos 1 lb por debajo del peso de cargo extra?

```python
# Distribución normal
mu = 12
sigma = 3.5
p = 0.99

print(f"X~N(mu={mu}, sigma²={sigma**2})")
print("X:=Peso de un paquete (lb)\n")

# P(X <= L - 1) = 0.99
# Primero buscamos x tal que F(x) = 0.99
# Después sumamos 1 lb para obtener el límite L

percentil_99 = brentq(lambda x: dist.norm.cdf(x, loc=mu, scale=sigma) - p, mu, mu + 6*sigma)
limite = percentil_99 + 1

probabilidad = dist.norm.cdf(limite - 1, loc=mu, scale=sigma)

print(f"Percentil 99 = {percentil_99:.4f} lb")
print(f"Peso para cargo extra = {limite:.4f} lb")
print(f"P(X <= {limite - 1:.4f}) = {probabilidad:.6f}")
```

### 19. La dispersión de las atomizaciones de los pesticidas es una preocupación constante de los fumi- gadores y los productores agrícolas. La relación inversa entre el tamaño de la gota y el potencial de deriva es bien conocida. El artículo “Effects of 2,4-D Formulation and Quinclorac on Spray Droplet Size and Deposition” (Weed Technology, 2005: 1030–1036) es una investigación sobre los efectos de las formulaciones de herbicidas en atomizaciones. Una figura en el artículo sugiere que la distribución normal con media de 1050 mm y desviación estándar de 150 mm es un modelo razonable del tamaño de las gotas de agua (el “tratamiento de control”) rociadas a través de una boquilla de 760 ml/min.
#### a) ¿Cuál es la probabilidad de que el tamaño de una sola gota sea de menos de 1500 m? ¿y al menos de 1000 m?
#### b) ¿Cuál es la probabilidad de que el tamaño de una sola gota este entre 1000 y 1500 mm?

```python
# Distribución normal
# Tamaño de las gotas de agua en mm
mu = 1050
sigma = 150

print(f"X~N(mu={mu}, sigma²={sigma**2})")
print("X:=Tamaño de una gota de agua (mm)\n")

# a)
# P(X < 1500) = P(X <= 1500) = F(1500)
a1 = dist.norm.cdf(1500, loc=mu, scale=sigma)

# P(X >= 1000) = 1 - P(X < 1000) = 1 - F(1000)
a2 = 1 - dist.norm.cdf(1000, loc=mu, scale=sigma)

# b)
# P(1000 <= X <= 1500) = F(1500) - F(1000)
b = (dist.norm.cdf(1500, loc=mu, scale=sigma) - dist.norm.cdf(1000, loc=mu, scale=sigma))

print(f"a1) P(X < 1500) = {a1:.6f}")
print(f"a2) P(X >= 1000) = {a2:.6f}")
print(f"b) P(1000 <= X <= 1500) = {b:.6f}")
```

### 20. En un proceso de pavimentación de una carretera la mezcla bituminosa se entrega en la tolva de la pavimentadora mediante camiones que transportan el material desde la planta de concreto. El artículo “Modeling of Simultaneously Continuous and Stochastic Construction Activities for Simulation” (J. of Construction Engr. and Mgmnt., 2013: 1037–1045) propone una distribución normal con media de 8.46 min y desviación estándar de 0.913 min para la variable aleatoria X tiempo del camión transportador.
#### a)¿Cuál es la probabilidad de que el tiempo de entrega del camión transportador sea al menos de 10 minutos? ¿Superará los 10 min?
#### b) ¿Cuál es la probabilidad de que el tiempo de entrega del camión transportador supere los 15 min?
#### c)¿ Cuál es la probabilidad de que el tiempo del camión transportador esté entre 8 y 10 min?

```python
# Distribución normal
# Tiempo de transporte del camión en minutos
mu = 8.46
sigma = 0.913

print(f"X~N(mu={mu}, sigma²={sigma**2:.6f})")
print("X:=Tiempo de transporte del camión (minutos)\n")

# a)
# P(X >= 10) = 1 - P(X < 10) = 1 - F(10)
# P(X > 10) = P(X >= 10), por ser continua
a = 1 - dist.norm.cdf(10, loc=mu, scale=sigma)

# b)
# P(X > 15) = 1 - F(15)
# Por simetría de la normal:
# P(X > 15) = F_Z((mu - 15) / sigma)
b = dist.norm.cdf((mu - 15) / sigma)

# c)
# P(8 <= X <= 10) = F(10) - F(8)
c = (
    dist.norm.cdf(10, loc=mu, scale=sigma) - dist.norm.cdf(8, loc=mu, scale=sigma))

print(f"P(X >= 10) = {a:.6f}")
print(f"P(X > 10) = {a:.6f}")
print(f"P(X > 15) = {b:.6e}")
print(f"P(8 <= X <= 10) = {c:.6f}")
```
