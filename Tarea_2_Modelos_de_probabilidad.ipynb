```python
from math import comb, exp, factorial

def binomial(n, p, x):
    return comb(n, x) * p**x * (1-p)**(n-x)

def poisson(lam, x):
    return exp(-lam) * lam**x / factorial(x)

def mostrar(prob):
    print(f"Probabilidad: {prob:.10f} ({prob:.6%})")
```

## Ejercicio 1
El 53 % de los trabajadores de una empresa son mujeres. Se seleccionan 8 personas al azar.

**Distribución binomial.** Sea $X$ el número de mujeres seleccionadas. Suponemos ensayos independientes con probabilidad constante: $n=8$, $p=0.53$, $q=0.47$.

$$P(X=x)=\binom{n}{x}p^xq^{n-x}$$

**a) Alguna mujer.** Por complemento:
$$P(X\geq1)=1-P(X=0)=1-(0.47)^8.$$

**b) Más de 6 mujeres.**
$$P(X>6)=P(X=7)+P(X=8)$$
$$=8(0.53)^7(0.47)+(0.53)^8.$$

```python
n, p = 8, 0.53
print("a) Alguna mujer:")
mostrar(1-binomial(n,p,0))
print("b) Más de 6 mujeres:")
mostrar(sum(binomial(n,p,x) for x in (7,8)))
```

## Ejercicio 2
La probabilidad de que un juguete salga defectuoso es 0.03. Se analiza un lote de 60 juguetes.

**Distribución binomial.** $X$ representa el número de juguetes defectuosos; $n=60$, $p=0.03$, $q=0.97$.

**a) Alguno defectuoso.**
$$P(X\geq1)=1-P(X=0)=1-(0.97)^{60}.$$

**b) Menos de dos defectuosos.**
$$P(X<2)=P(X=0)+P(X=1)$$
$$=(0.97)^{60}+60(0.03)(0.97)^{59}.$$

```python
n, p = 60, 0.03
print("a) Alguno defectuoso:")
mostrar(1-binomial(n,p,0))
print("b) Menos de dos defectuosos:")
mostrar(sum(binomial(n,p,x) for x in range(2)))
```

## Ejercicio 3
La probabilidad de una reacción positiva en un examen médico es 0.8. Se realizan 7 pruebas y se busca la probabilidad de menos de 5 reacciones negativas.

**Distribución binomial.** $X$ es el número de reacciones negativas; $n=7$, $p=1-0.8=0.2$, $q=0.8$.

$$P(X<5)=1-P(X\geq5)$$
$$=1-\left[\binom75(0.2)^5(0.8)^2+\binom76(0.2)^6(0.8)+\binom77(0.2)^7\right]$$
$$=1-(0.0043008+0.0003584+0.0000128).$$

```python
n, p = 7, 0.2
mostrar(1-sum(binomial(n,p,x) for x in range(5,8)))
```

## Ejercicio 4
Un hospital recibe en promedio 43.2 pacientes de urgencias durante 24 horas. El servicio colapsa si recibe más de 50 pacientes.

**Distribución de Poisson.** Sea $X$ el número de pacientes en 24 horas y $\lambda=43.2$.

$$P(X=x)=\frac{e^{-\lambda}\lambda^x}{x!}$$

Por complemento:
$$P(X>50)=1-P(X\leq50)$$
$$=1-\sum_{x=0}^{50}\frac{e^{-43.2}(43.2)^x}{x!}.$$

```python
lam = 43.2
mostrar(1-sum(poisson(lam,x) for x in range(51)))
```

## Ejercicio 5
Un detergente lava correctamente el 88 % de las prendas. Se lavan 10 prendas.

**Distribución binomial.** $n=10$.

**a) Buenos resultados en las 10 prendas.** Sea $X$ el número de prendas bien lavadas, con $p=0.88$.
$$P(X=10)=\binom{10}{10}(0.88)^{10}(0.12)^0=(0.88)^{10}.$$

**b) Al menos 2 prendas mal lavadas.** Sea $Y$ el número de prendas mal lavadas, con $p=0.12$.
$$P(Y\geq2)=1-[P(Y=0)+P(Y=1)]$$
$$=1-[(0.88)^{10}+10(0.12)(0.88)^9].$$

```python
n = 10
print("a) Las 10 prendas bien lavadas:")
mostrar(binomial(n,0.88,10))
print("b) Al menos 2 prendas mal lavadas:")
mostrar(1-sum(binomial(n,0.12,x) for x in range(2)))
```

## Ejercicio 6
El 3 % de las declaraciones fiscales presentan errores. Se considera un grupo de 10 declaraciones.

**Distribución binomial.** Sea $X$ el número de declaraciones con errores; $n=10$, $p=0.03$, $q=0.97$.

$$P(X<5)=P(X\leq4)=\sum_{x=0}^{4}\binom{10}{x}(0.03)^x(0.97)^{10-x}$$
$$=(0.97)^{10}+10(0.03)(0.97)^9+45(0.03)^2(0.97)^8$$
$$\quad+120(0.03)^3(0.97)^7+210(0.03)^4(0.97)^6.$$

**Nota:** El enunciado también menciona inspeccionar hasta encontrar un error (modelo geométrico). Se interpreta la pregunta principal como el número de declaraciones con errores en un grupo fijo de 10. Si se pidiera encontrar el primer error antes de la quinta inspección, la probabilidad sería $1-(0.97)^4$.

```python
n, p = 10, 0.03
mostrar(sum(binomial(n,p,x) for x in range(5)))
```

## Ejercicio 7
El número de componentes que fallan sigue una distribución de Poisson, con un promedio de 8 fallos en 100 horas.

**Distribución de Poisson.** Suponemos tasa constante $r=8/100=0.08$ fallos por hora.

**a) Exactamente un fallo en 25 horas.**
$$\lambda=0.08(25)=2$$
$$P(X=1)=\frac{e^{-2}2^1}{1!}=2e^{-2}.$$

**b) No más de dos fallos en 50 horas.**
$$\lambda=0.08(50)=4$$
$$P(X\leq2)=e^{-4}\left(1+4+\frac{4^2}{2!}\right)=13e^{-4}.$$

```python
r = 8/100
print("a) Un fallo en 25 horas:")
mostrar(poisson(r*25,1))
print("b) No más de dos fallos en 50 horas:")
mostrar(sum(poisson(r*50,x) for x in range(3)))
```

## Ejercicio 8
Los mensajes llegan a un servidor según una distribución de Poisson con tasa promedio de 0.1 mensajes por minuto. Se busca la probabilidad de recibir como máximo dos mensajes en una hora.

**Distribución de Poisson.** Sea $X$ el número de mensajes en 60 minutos.
$$\lambda=0.1(60)=6$$
$$P(X\leq2)=P(X=0)+P(X=1)+P(X=2)$$
$$=e^{-6}\left(1+6+\frac{6^2}{2!}\right)=25e^{-6}.$$

```python
lam = 0.1*60
mostrar(sum(poisson(lam,x) for x in range(3)))
```

## Ejercicio 9
El 5 % de los pañuelos producidos por una empresa presenta algún defecto. Se empaquetan en cajas de 15 unidades.

**Distribución binomial.** Sea $X$ el número de pañuelos defectuosos en una caja; $n=15$, $p=0.05$, $q=0.95$.

**a) Exactamente dos defectuosos.**
$$\binom{15}{2}=\frac{15(14)}{2}=105$$
$$P(X=2)=105(0.05)^2(0.95)^{13}.$$

**b) Menos de tres defectuosos.**
$$P(X<3)=P(X=0)+P(X=1)+P(X=2)$$
$$=(0.95)^{15}+15(0.05)(0.95)^{14}+105(0.05)^2(0.95)^{13}.$$

```python
n, p = 15, 0.05
print("a) Exactamente dos defectuosos:")
mostrar(binomial(n,p,2))
print("b) Menos de tres defectuosos:")
mostrar(sum(binomial(n,p,x) for x in range(3)))
```
